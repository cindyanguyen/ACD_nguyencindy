Following the important work in cultural analytics, computational hermeneutics, and computational humanities, what might a critical computational history look like? The computational turn in historical methodologies and research practices have already arrived expedited by the increasing amounts of digitized and accessible corpora, several decades of important digital history research in mapping and visualization, and the influence of computational methods from machine learning, computer science, and artificial intelligence. While previous digital historical work honed into specific digital workflows for historical questions, a wider more interoperable research framework is needed in order to communicate how large scale computational analysis might inform historical research stages that could be pedagogically meaningful across research projects and into other disciplines. Drawing from core concepts of ‘task abstraction’ from data visualization and computer science framework (Munzner, 2015), this article hones into the specific domain specific research ‘tasks’ that historians (users) seek to do through their primary sources (data), and moves towards communicating the ‘abstracted tasks’ for the purpose of reproducibility in computational historical work. Focused on the computational tasks of data inquiry, context, and historical argument, it poses specific examples where machine learning and artificial intelligence might inform specific stages of historical inquiry, warning against the current trend towards wholesale adaptation of computational work for every stage. Furthermore, it addresses the fundamental shape of interpretability, validity, and evaluation within computational historical work, and raises questions of what parts of collaborative, long term historical research projects might move towards reproducibility versus transparency. By honing into the ways in which historical knowledge is produced, it also addresses certain disciplinary differences in inductive and deductive reasoning that lend itself to different types of computational tasks.

Ultimately this work recenters the work of computational history within the ‘evergreen’ technology of teaching and implementing core practices of research design and project management. This article builds on the fundamental work of exploratory data visualization in historical research (Sanders, 2024) and more recent work on how exploratory computation (Nguyen and Alvarado Rojas, 2026) could facilitate interpretive and inductive reasoning tasks. Furthermore, it addresses core problems within computational historical tasks, such as the uneven digitized cultural record (Risam, 2019) as well as how to address historical bias in problematic sources, many of which were gathered or preserved within the context of colonialism, extraction, and violence. This article warns against the reproduction of hegemonic forms of representation and analysis that is further escalated if computational methodologies are used uncritically, such as seen in reproducing biases and absences in training data or reproducing mis/disinformation through the hallucinating tendencies of generative artificial intelligence. Drawing from fields such as postcolonial and critical computing, it advocates for an approach to computational historical methodology that both capaciously embraces multivocal knowledgemaking across disciplines, methodologies, and corpora and transparently communicates limitations within the research design. In conversation with decolonial and feminist theory, Indigenous and Southeast Asian studies, this paper argues that a critical interpretive digital humanities can make visible the layers of colonial representation, surface plural voices and multidirectional narratives, and dwell in the discomforts of unknowability. This commitment seeks to challenge top down, linear, hegemonic colonial representations that often materialize in static forms of data representations, the privileging of Western frameworks of authorship, and a scientific push towards empirical argumentative findings. As a provocation on research praxis, this paper and talk confronts disciplinary global north structures of knowledge production and advocates for multivocality as research design principle.
